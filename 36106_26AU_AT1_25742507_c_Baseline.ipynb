## A. Assess Baseline Model

```python
# DO NOT MODIFY THE CODE IN THIS CELL
# Load data
try:
  X_train = pd.read_csv(at.folder_path / 'X_train.csv')
  y_train = pd.read_csv(at.folder_path / 'y_train.csv')

  X_val = pd.read_csv(at.folder_path / 'X_val.csv')
  y_val = pd.read_csv(at.folder_path / 'y_val.csv')

  X_test = pd.read_csv(at.folder_path / 'X_test.csv')
  y_test = pd.read_csv(at.folder_path / 'y_test.csv')
except Exception as e:
  print(e)
```

### A.1 Generate Predictions with Baseline Model

```python
baseline_prediction = y_train.mean().values[0]

y_train_pred_baseline = np.full(len(y_train), baseline_prediction)
y_val_pred_baseline = np.full(len(y_val), baseline_prediction)
y_test_pred_baseline = np.full(len(y_test), baseline_prediction)
```

### A.2 Selection of Performance Metrics

> Provide some explanations on why you believe the performance metrics you chose is appropriate


```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mae = mean_absolute_error(y_val, y_val_pred_baseline)
rmse = np.sqrt(mean_squared_error(y_val, y_val_pred_baseline))
r2 = r2_score(y_val, y_val_pred_baseline)

print("MAE:", mae)
print("RMSE:", rmse)
print("R2:", r2)
```
