## C. Train Machine Learning Model

### C.1 Import Algorithm



```python
from sklearn.linear_model import LinearRegression
```

```python
algorithm_selection_explanations = "I selected linear regression as the first machine learning model because it is one of the most fundamental and interpretable algorithms for regression problems. It models the relationship between the input features and the target variable as a linear combination of the predictors. In this context, linear regression is appropriate because vehicle price is expected to be influenced by many numerical and categorical features such as manufacturing year, engine capacity, fuel consumption, and kilometres driven. These relationships can be approximated by linear trends.Using this as the first experiment also provides a useful benchmark for understanding how well the selected features explain the variability in vehicle prices before moving to more complex models."
```

```python
# DO NOT MODIFY THE CODE IN THIS CELL
print_tile(size="h3", key='algorithm_selection_explanations', value=algorithm_selection_explanations)
```

```python
X_train = X_train.fillna(X_train.median(numeric_only=True))
X_val = X_val.fillna(X_train.median(numeric_only=True))
X_test = X_test.fillna(X_train.median(numeric_only=True))
```

```python
X_train = pd.get_dummies(X_train, drop_first=True)
X_val = pd.get_dummies(X_val, drop_first=True)
X_test = pd.get_dummies(X_test, drop_first=True)

X_val = X_val.reindex(columns=X_train.columns, fill_value=0)
X_test = X_test.reindex(columns=X_train.columns, fill_value=0)
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)
```

Before scaling the data, categorical variables were transformed using one-hot encoding.
Machine learning algorithms require numerical input, so categorical features such as vehicle brand, vehicle type, fuel type, and territory were converted into binary variables representing each category.

After encoding the categorical variables, feature scaling was applied using StandardScaler. Scaling standardizes the numerical variables so that they have a mean of zero and a standard  deviation of one. This step is important for linear models because it ensures that variables with larger numerical ranges do not dominate the regression coefficients.

### C.2 Set Hyperparameters


```python
# First model: Linear regression with intercept
model_intercept = LinearRegression(fit_intercept=True)

# Second model: Linear regression without intercept
model_no_intercept = LinearRegression(fit_intercept=False)
```

```python
hyperparameters_selection_explanations = "The hyperparameter tested in this experiment is fit_intercept within linear regression model. This parameter determines whether the regression model should estimate an intercept term in the regression equation. When fit_intercept=True, the model learns a baseline value representing the expected vehicle price when all predictor variables are zero. When fit_intercept=False, the regression line is forced to pass through the origin (0,0). Tuning this hyperparameter allows us to evaluate whether including an intercept improves the predictive performance of the model. In the context of pricing problems, it is unlikely that the price of a vehicle would naturally be zero even if the predictor variables were zero. That's why allowing the model to estimate an intercept could better capture the relationship between the features and price.Testing both model variations helps determining which one produces more accurate predictions and provides insight into how the regression model represents the pricing structure."
```

```python
# DO NOT MODIFY THE CODE IN THIS CELL
print_tile(size="h3", key='hyperparameters_selection_explanations', value=hyperparameters_selection_explanations)
```

### C.3 Fit Model

```python
# Train model with intercept
model_intercept.fit(X_train_scaled, y_train)

# Train model without intercept
model_no_intercept.fit(X_train_scaled, y_train)
```

---
## D. Model Evaluation

### D.1 Model Technical Performance

```python
# Predictions of each sub-model
# Predictions with intercept
y_pred_intercept = model_intercept.predict(X_val_scaled)

# Predictions without intercept
y_pred_no_intercept = model_no_intercept.predict(X_val_scaled)
```

```python
# Model with intercept
mae_intercept = mean_absolute_error(y_val, y_pred_intercept)
rmse_intercept = np.sqrt(mean_squared_error(y_val, y_pred_intercept))
r2_intercept = r2_score(y_val, y_pred_intercept)

# Model without intercept
mae_no_intercept = mean_absolute_error(y_val, y_pred_no_intercept)
rmse_no_intercept = np.sqrt(mean_squared_error(y_val, y_pred_no_intercept))
r2_no_intercept = r2_score(y_val, y_pred_no_intercept)

print("Model with intercept")
print("MAE:", mae_intercept)
print("RMSE:", rmse_intercept)
print("R2:", r2_intercept)

print("\nModel without intercept")
print("MAE:", mae_no_intercept)
print("RMSE:", rmse_no_intercept)
print("R2:", r2_no_intercept)
```

```python
model_performance_explanations = "These results show a clear difference between the two models. The model with fit_intercept=True performs much better than the model without the intercept. The MAE for the model with intercept is about 9,636, which means that on average the predicted car prices differ from the real prices by around $9,600 AUD. The RMSE is about 21,214, which is higher because this metric penalizes larger errors more strongly. The R2 value is 0.57, meaning that the model explains about 57% of the variation in vehicle prices. This indicates that the features I selected are able to capture a reasonable part of the pricing patterns in the dataset. On the other hand, the model without the intercept performs very bad. The MAE is larger than 31,000, which means the predictions are much further away from the real prices. The RMSE is also much higher, and the R2 value is negative (-0.30), and as we know, a negative R2 means that the model is performing worse than a very simple model that would just predict the average price for every car. This could be happening because forcing the regression line to pass through zero (0,0) does not reflect the real relationship between the vehicle features and the price. In summary, these results suggest that including an intercept is necessary for this problem. Car prices do not naturally start from zero, so allowing the model to estimate a baseline value makes the predictions much more realistic and accurate."
```

```python
# DO NOT MODIFY THE CODE IN THIS CELL
print_tile(size="h3", key='model_performance_explanations', value=model_performance_explanations)
```
