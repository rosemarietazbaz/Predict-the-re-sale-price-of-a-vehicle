## C. Train Machine Learning Model

### C.1 Import Algorithm



```python
from sklearn.neighbors import KNeighborsRegressor
```

```python
# <Student to fill this section and then remove this comment>
algorithm_selection_explanations = " In this experiment, I use the KNN regression algorithm to predict vehicle resale prices. Unlike linear regression models, KNN predicts the price of a vehicle by looking at the prices of the most similar vehicles in the dataset. This approach may be suitable for the problem because vehicle prices often depend on combinations of features such as brand, mileage, engine capacity, and vehicle type. KNN can capture these relationships by comparing each vehicle with similar vehicles that already exist in the dataset. From a business perspective, this approach is intuitive because it estimates a vehicle’s price based on the prices of comparable vehicles in the market."
```

```python
# DO NOT MODIFY THE CODE IN THIS CELL
print_tile(size="h3", key='algorithm_selection_explanations', value=algorithm_selection_explanations)
```

```python
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.impute import SimpleImputer

# Copies
X_train_enc = X_train.copy()
X_val_enc = X_val.copy()
X_test_enc = X_test.copy()

# Encode categorical variables
encoders = {}

categorical_cols = X_train_enc.select_dtypes(include="object").columns

for col in categorical_cols:

    le = LabelEncoder()

    combined_values = pd.concat([
        X_train_enc[col],
        X_val_enc[col],
        X_test_enc[col]
    ])

    le.fit(combined_values.astype(str))

    X_train_enc[col] = le.transform(X_train_enc[col].astype(str))
    X_val_enc[col] = le.transform(X_val_enc[col].astype(str))
    X_test_enc[col] = le.transform(X_test_enc[col].astype(str))

    encoders[col] = le

# missing values
imputer = SimpleImputer(strategy="median")

X_train_enc = imputer.fit_transform(X_train_enc)
X_val_enc = imputer.transform(X_val_enc)
X_test_enc = imputer.transform(X_test_enc)


# Scaling
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_enc)
X_val_scaled = scaler.transform(X_val_enc)
X_test_scaled = scaler.transform(X_test_enc)
```

In this step, the dataset was prepared so it could be used by the KNN algorithm. KNN calculates distances between observations, so all features must be numeric and scaled.

Categorical variables such as vehicle_brand, model_name, vehicle_type, fuel_type, transmission_type, drive_type, vehicle_colour, and territory were converted into numerical values using LabelEncoder.

During this step, it was also necessary to ensure that the encoder could handle categories appearing in the validation or testing datasets that were not present in the training dataset. For example, some vehicle brands appeared only in the validation set. Since LabelEncoder cannot transform categories that it has not seen before, the encoder was fitted using the combined values from the training, validation, and testing datasets. This ensures that all possible categories are recognised and avoids transformation errors during model evaluation.

After encoding the categorical variables, the dataset was scaled using StandardScaler. Scaling is very important for the KNN algorithm because it relies on distance calculations between observations. If variables are on very different scales, features with larger numeric ranges could dominate the distance calculation and negatively affect the model’s predictions.

By standardizing the variables, each feature contributes more equally to the similarity calculations, allowing the KNN model to identify similar vehicles more effectively and generate more reliable price predictions.

### C.2 Set Hyperparameters


```python
# model 1
knn_model_1 = KNeighborsRegressor(
    n_neighbors=5,
    weights='uniform',
    metric='minkowski'
)

# model 2
knn_model_2 = KNeighborsRegressor(
    n_neighbors=10,
    weights='distance',
    metric='minkowski'
)
```

```python
# <Student to fill this section and then remove this comment>
hyperparameters_selection_explanations = "In this experiment, I used different hyperparameter values, to see how they affect the prediction performance of the KNN model. The main hyperparameter tuned is n_neighbors, which defines how many similar vehicles are used to estimate the resale price. A smaller number of neighbours focuses on the most similar vehicles, which may capture more specific patterns in the data. A larger number of neighbours produces smoother predictions because it averages over more observations. The weights parameter determines how the neighbours influence the final prediction. With uniform weights, all neighbours contribute equally to the predicted price. With distance weights, closer vehicles have more influence on the prediction than those that are further away. The distance metric used is Minkowski distance (standard Euclidean distance when p = 2). Testing different values for these hyperparameters allows us to evaluate how sensitive the model is to the number of neighbours and how similarity between vehicles influences the predicted price. This comparison helps identify the configuration that produces the lowest prediction error for the vehicle pricing problem."
```

```python
# DO NOT MODIFY THE CODE IN THIS CELL
print_tile(size="h3", key='hyperparameters_selection_explanations', value=hyperparameters_selection_explanations)
```

### C.3 Fit Model

```python
knn_model_1.fit(X_train_scaled, y_train.values.ravel())
knn_model_2.fit(X_train_scaled, y_train.values.ravel())
```

## D. Model Evaluation

### D.1 Model Technical Performance

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Predictions
y_pred_knn_1 = knn_model_1.predict(X_val_scaled)
y_pred_knn_2 = knn_model_2.predict(X_val_scaled)

# Convert y_val to array
y_val_array = y_val.values.ravel()

# 1
mae_knn_1 = mean_absolute_error(y_val_array, y_pred_knn_1)
rmse_knn_1 = np.sqrt(mean_squared_error(y_val_array, y_pred_knn_1))
r2_knn_1 = r2_score(y_val_array, y_pred_knn_1)

print("KNN Model 1")
print("MAE:", mae_knn_1)
print("RMSE:", rmse_knn_1)
print("R2:", r2_knn_1)

print("\n")

# 2
mae_knn_2 = mean_absolute_error(y_val_array, y_pred_knn_2)
rmse_knn_2 = np.sqrt(mean_squared_error(y_val_array, y_pred_knn_2))
r2_knn_2 = r2_score(y_val_array, y_pred_knn_2)

print("KNN Model 2")
print("MAE:", mae_knn_2)
print("RMSE:", rmse_knn_2)
print("R2:", r2_knn_2)
```

```python
# <Student to fill this section and then remove this comment>
model_performance_explanations = "The results show that KNN model 1 performs slightly better than KNN model 2. KNN Model 1 achieved a MAE of about $11,750, meaning that on average the predicted vehicle price differs from the validation price by approximately $11,000. The RMSE is around $25,763, which represents the typical prediction error when larger mistakes are penalized more heavily. The R2 value is 0.36, indicating that the model is able to explain about 36% of the variation in vehicle prices in the validation dataset. KNN model 2 performs slightly worse. It has a MAE of around $11,878 and an RMSE of about $26,221, meaning the prediction errors are slightly larger. Its R2 value is 0.34, showing that it explains slightly less of the variation in vehicle prices compared to Model 1. Since RMSE is the most important metric for this project, model 1 is considered the better KNN configuration because it produces the lowest prediction error. However, when comparing these results with the previous experiments, the KNN models perform worse than both linear regression and elastic net. In particular, the RMSE obtained with KNN is significantly higher, which means that the price predictions are less accurate. From a modelling perspective, this suggests that the relationship between features and resale price in this dataset may be captured more effectively by linear models than by similarity-based approaches like KNN. Because KNN relies heavily on the distance between observations, the presence of many categorical variables and potentially high variation between similar vehicles may reduce its capacity to identify reliable neighbours. Overall, while KNN can still provide general price estimates, its higher RMSE indicates that it is less suitable for this pricing problem compared to the models tested in the previous experiments."
```
